# **Módulo de autenticación de usuarios**
## ***Integrante 1 - Implementación del módulo***

```python

# Credenciales ficticias para el laboratorio
usuarios = {
    "admin": "Admin123",
    "estudiante": "Estudiante123",
    "invitado": "Invitado123"
}

# Función de autenticación
def autenticar(usuario, contrasena):

    if usuario == "" or contrasena == "":
        return "Campos vacíos"

    if usuario not in usuarios:
        return "Usuario no encontrado"

    if usuarios[usuario] != contrasena:
        return "Contraseña incorrecta"

    return "Acceso permitido"
```

## **Criterios de evaluación de confiabilidad**

- *Ejecución exitosa:* el resultado obtenido coincide
  con el resultado esperado.

- *Ejecución fallida:* el resultado obtenido es diferente
  al esperado o se produce un error de ejecución.

Un acceso rechazado correctamente también se considera
una ejecución exitosa.

## **Pruebas iniciales del módulo de autenticación**

```python
# Pruebas iniciales del módulo
pruebas = [
    ("admin", "Admin123", "Acceso permitido"),
    ("admin", "12345", "Contraseña incorrecta"),
    ("desconocido", "12345", "Usuario no encontrado"),
    ("", "Admin123", "Campos vacíos"),
    ("admin", "", "Campos vacíos"),
    ("estudiante", "Estudiante123", "Acceso permitido"),
    ("invitado", "Invitado123", "Acceso permitido")
]

for i, (usuario, contrasena, esperado) in enumerate(pruebas, 1):
    obtenido = autenticar(usuario, contrasena)
    estado = "EXITOSA" if obtenido == esperado else "FALLIDA"

    print(f"Prueba {i}")
    print(f"Usuario: {usuario}")
    print(f"Resultado esperado: {esperado}")
    print(f"Resultado obtenido: {obtenido}")
    print(f"Estado: {estado}")
    print("-" * 45)
```
